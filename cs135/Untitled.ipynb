```python
Pass=input("Enter your password :")
n=4

for i in range(0,n+1):
    if(i==n):
        print("Try after 2 minutes")
        break
    Reenter=input("Enter your password for checking")
    if(Pass == Reenter):
        print("Password matched logined")
        break
    else:
        print(f"Try again !! chances remaining {n-i} times")

```
